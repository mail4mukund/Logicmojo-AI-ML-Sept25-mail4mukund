Basic Lambda function syntax
lambda args : arg1 * arg2 * arg3;

```python
x=100;
result=lambda x: x*x;
print(result(x))
```

```python
#if condigiotn in one line use lambda function
lm= lambda x, y:"More than hundred" if (x+y)>100  else "Less than hundred";
print(lm(100,11));
print(lm(10,11));
```

```python
#lambda with Filter, Map, Reduce, Sorted methods usage
x = [aa for aa in range(0,100)]
even=list(filter(lambda oe:oe%2==0,x))
print(len(even))
for k in even:
    print(k)
```
