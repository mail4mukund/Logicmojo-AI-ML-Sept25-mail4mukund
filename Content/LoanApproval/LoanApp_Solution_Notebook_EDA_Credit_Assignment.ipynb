###### We would use a logarthmic scale since the values as we saw above are huge and a function to plot all the graph which takes into the arguments the axis name and datasets to plot the required graph, In this case a bar graph

```python
# Count plotting in logarithmic scale

def cplot(df,col,title,hue =None):

    plt.rcParams["axes.labelsize"] = 20
    plt.rcParams['axes.titlesize'] = 20
    plt.rcParams['axes.titlepad'] = 20

    plt.figure(figsize=(50,10))
    temp = pd.Series(data = hue)
    fig, ax = plt.subplots()
    plt.xticks(rotation=90)
    plt.yscale('log')
    plt.title(title)
    ax = sns.countplot(data = df, x= col, hue = hue,palette='mako')



    plt.show()
```
